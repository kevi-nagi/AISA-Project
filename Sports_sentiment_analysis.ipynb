```python
import random
import matplotlib.pyplot as plt
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
import pandas as pd
```

```python
comments_base = [
    "Messi wins it in the 94th minute! 🐐", "I can't believe that free kick! Unreal.",
    "Bro... Messi still got it at 38 🔥", "GOAT does it again, what a player!",
    "Crowd reaction gave me chills 😍", "Just wow. No words. Messi magic.",
    "Messi’s goal was pure class.", "This man is not human 😭",
    "Another last-minute winner 😱", "Still the greatest of all time.",
    "That goal was poetry in motion.", "The way he bent that shot 😍🔥",
    "Messi showing why he's the GOAT again.", "Unreal technique. Goal of the season.",
    "Messi just made history again!", "Vintage Messi performance tonight.",
    "He plays like he's still 25. Amazing.", "Clutch gene activated!",
    "His vision and control is unmatched.", "We are lucky to witness this. 🙌",
    "Game was mid tbh. Just one moment.", "Amazing finish to a boring match.",
    "It was a close match overall.", "Galaxy had their chances too.",
    "Messi scored again. No surprise.", "It ended 2-1 in favor of Miami.",
    "The match had its ups and downs.", "The crowd was loud tonight.",
    "It was a typical MLS game.", "Both teams looked tired.",
    "Ref gave everything to Inter Miami 😒", "LA Galaxy robbed. VAR was a joke.",
    "Refs were blind all night. Disgrace.", "Tired of this league rigging everything.",
    "Can someone check the VAR refs please?", "League scripted for Messi lol.",
    "Match was ruined by officiating.", "Always the same bias for Messi.",
    "LA played better but still lost. Unfair.", "This ref should never officiate again.",

]

# Adding fillers to reach 200 comments total
while len(comments_base) < 200:
    comments_base.append(f"Random fan comment number {len(comments_base)+1}")
```

```python
analyzer = SentimentIntensityAnalyzer()

sentiments = []
for comment in comments_base:
    score = analyzer.polarity_scores(comment)['compound']
    if score >= 0.05:
        sentiments.append('positive')
    elif score <= -0.05:
        sentiments.append('negative')
    else:
        sentiments.append('neutral')
```

```python
df = pd.DataFrame({
    'Comment': comments_base,
    'Sentiment': sentiments
})
```

```python
sentiment_counts = df['Sentiment'].value_counts()
```

```python
plt.figure(figsize=(8, 4))
sentiment_counts.plot(kind='bar', color=['lightcoral', 'blue', 'red'])
plt.title('Fan Sentiment Distribution (Bar Chart)')
plt.xlabel('Sentiment')
plt.ylabel('Number of Comments')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6, 6))
colors = ['lightcoral','lightgreen', 'lightpink']
sentiment_counts.plot.pie(autopct='%1.1f%%', startangle=90, colors=colors, ylabel='')
plt.title('Fan Sentiment Distribution (Pie Chart)')
plt.show()
```
