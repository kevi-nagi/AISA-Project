```python
import numpy as np
```

```python
FEATURE_TAGS = ['Goal', 'Save', 'Foul']
```

```python
HIGHLIGHTS = {
    'H1: LeBron Dunk': [1, 0, 0],
    'H2: Messi Foul': [0, 0, 1],
    'H3: Haaland Goal': [1, 0, 0],
    'H4: Djokovic Save': [0, 1, 0],
    'H5: Curry 3-Pointer': [1, 0, 0],
    'H6: Jokic Block': [0, 1, 0],
    'H7: Crosby Goal': [1, 0, 0],
    'H8: Alisson Save': [0, 1, 0],
    'H9: Serena Foul': [0, 0, 1],
    'H10: Giannis Foul': [0, 0, 1],
    'H11: Ohtani HR': [1, 0, 0],
    'H12: McDavid Save': [0, 1, 0],
    'H13: Cole Strikeout': [0, 1, 0],
    'H14: Neymar Dive': [0, 0, 1],
    'H15: Nadal Winner': [1, 0, 0],
}
```

```python
highlight_names = list(HIGHLIGHTS.keys())
highlight_vectors = np.array(list(HIGHLIGHTS.values()))
```

```python
FANS = {
    'John': [5, 3, 1],
    'Rose': [2, 5, 4],
    'Violet': [4, 4, 2],
    'Jude': [1, 2, 5],
    'Kenji': [5, 1, 2],
}
```

```python
def generate_recommendations(fans, highlights, names):

    all_recommendations = {}

    for fan_name, fan_vector_list in fans.items():
        fan_vector = np.array(fan_vector_list)

        scores = np.dot(highlight_vectors, fan_vector)

        scored_highlights = list(zip(names, scores))

        scored_highlights.sort(key=lambda x: x[1], reverse=True)

        top_3_recommendations = scored_highlights[:3]
        all_recommendations[fan_name] = top_3_recommendations

    return all_recommendations, scored_highlights
```

```python
if __name__ == "__main__":

    recommendations, all_scores = generate_recommendations(FANS, HIGHLIGHTS, highlight_names)

    print("PERSONALIZED HIGHLIGHT RECOMMENDATION SYSTEM")

    print("\n Raw Score Table (Score per Highlight Type)")

    fan_vectors = {name: vector for name, vector in FANS.items()}

    print("\n{:>8} | {:>5} | {:>5} | {:>5}".format("Fan Name", "Goal", "Save", "Foul"))
    print("-" * 33)

    for fan_name, vector in fan_vectors.items():
        score_goal = vector[0] * 1
        score_save = vector[1] * 1
        score_foul = vector[2] * 1
        print("{:>8} | {:>5} | {:>5} | {:>5}".format(fan_name, score_goal, score_save, score_foul))


    print("\n Top 3 Personalized Recommendations for Each Fan")

    for fan_name, top_clips in recommendations.items():
        print(f"\nRecommended for {fan_name}:")
        for i, (clip_name, score) in enumerate(top_clips):
            print(f"  {i+1}. {clip_name} (Score: {score:.1f})")
```
